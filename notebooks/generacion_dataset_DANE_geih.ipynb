```python
from pathlib import Path
import dane_data as dd
```

```python
CARPETA = Path("../data/raw")
SALIDA = CARPETA / "geih_2025_ocupados_crudo.csv"
MESES = None  # Meses a cargar, p. ej. [1, 2, 3] (None = todos)

# Variables del DANE que se guardan (códigos originales, sin traducir)
VARIABLES = [
    # llaves y diseño muestral
    "DIRECTORIO", "SECUENCIA_P", "ORDEN", "HOGAR", "MES", "DPTO", "CLASE", "FEX_C18",
    # persona
    "P6040",      # edad
    "P3271",      # sexo
    "P6050",      # parentesco con el jefe de hogar
    "P6070",      # estado civil
    "P3042",      # nivel educativo
    # empleo (módulo ocupados)
    "OCI",        # marca de ocupado
    "P6920",      # ¿cotiza a pensión?  <- objetivo
    "P6430",      # posición ocupacional
    "P3069",      # tamaño de la empresa (en 2025 reemplaza a P6870)
    "P6880",      # lugar donde trabaja
    "P6800",      # horas trabajadas a la semana
    "P6426",      # meses en el empleo
    "P6440",      # ¿tiene contrato?
    "INGLABO",    # ingreso laboral
    "RAMA2D_R4",  # rama de actividad
    # vivienda
    "P4030S1A1",  # estrato
]

CARPETA.mkdir(parents=True, exist_ok=True)
dd.set_paths(geih=str(CARPETA / "geih"))

print("Descargando GEIH 2025 del DANE (la primera vez tarda; es incremental)...")
fallidos = dd.geih.download(2025)
if fallidos:
    print(f"⚠️  No se pudo descargar: {fallidos}")

df = dd.geih.load(2025, months=MESES, columns=VARIABLES, harmonize_vars=False)
df = df[df["OCI"] == 1]  # solo ocupados (OCI se lee como número: 1.0)

faltan = [v for v in VARIABLES if v not in df.columns]
if faltan:
    print(f"⚠️  No vinieron estas variables (revisar nombre en el diccionario 2025): {faltan}")

cols = [c for c in ["anio", "mes", *VARIABLES] if c in df.columns]
df[cols].to_csv(SALIDA, index=False, encoding="utf-8")
print(f"Ocupados guardados: {len(df):,} filas, {len(cols)} columnas -> {SALIDA}")
```
